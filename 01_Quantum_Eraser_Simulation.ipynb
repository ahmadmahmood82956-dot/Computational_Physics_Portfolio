```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random

random.seed(42)
photonIDs=[]
screenPositions=[]
assignedDetectors=[]

detector_options=['D1','D2','D3','D4']

for count in range (1,1001):
    photonIDs.append(count)
    screenPositions.append(random.uniform(-4,4))
    assignedDetectors.append(random.choice(detector_options))

quantum_info = pd.DataFrame({
    'Photon_ID':photonIDs,
    'Screen_position':screenPositions,
    'Detector': assignedDetectors
})

particle_data= quantum_info[quantum_info['Detector'].isin(['D3','D4'])]
wave_data= quantum_info[quantum_info['Detector'].isin(['D1','D2'])]

plt.figure(figsize=(6,6))
sns.histplot(data=particle_data, x='Screen_position',bins=20,color='red')
plt.title('Detectors 3 and 4 : Known Path (particle behaviour)')
plt.xlabel('screen position cordinates')
plt.ylabel('number of photons')
plt.tight_layout()
plt.show()

plt.figure(figsize=(6,6))
sns.histplot(data=wave_data,x='Screen',bins=20,colour='blue')
plt.title('Detectors 1 and 2 : Erased Path (wave Interference)')
plt.xlabel('screen position cordinates')
plt.ylabel('number of photons')
plt.tight_layout()
plt.show()
```
