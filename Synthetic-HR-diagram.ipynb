```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
raw_data= pd.read_csv("https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Stars.csv")

# Standardize the 'Star color' column to ensure consistency with palette keys
raw_data['Star color'] = raw_data['Star color'].str.strip().str.title()

plt.style.use('dark_background')
plt.figure(figsize=(8,6), dpi=300)
custom_palette= {
    'Blue':'blue',
    'Red' : 'red',
    'White': 'white',
    'Blue White': 'royalblue',
    'Blue-White':'skyblue',
    'Orange': 'orange',
    'Orange-Red': 'orangered',
    'Whitish': 'whitesmoke',
    'Yellowish': 'yellow',
    'Yellowish White': 'lightyellow',
    'White-Yellow': 'lightgoldenrodyellow',
    'Yellow-White': 'lemonchiffon',
    'Pale Yellow Orange': 'moccasin'
    }
sns.scatterplot(data=raw_data,x="Temperature (K)",
                y="Luminosity (L/Lo)", size= "Radius (R/Ro)",
                hue="Star color",palette=custom_palette)
plt.gca().invert_xaxis()
plt.yscale('log')
plt.title('Hertzsprung-Russel(HR) Diagram (simulated data)')
plt.xlabel('Surface Temperature (K)')
plt.ylabel('Luminosity(Lo) -log scale')
plt.tight_layout()
plt.show()
```
